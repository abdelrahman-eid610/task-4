```python
import matplotlib.pyplot as plt 
import  numpy as np
x=np.array([10,9,26,40])
y=np.array([30,40,50,60])
plt.plot(x,y,linestyle="dotted")
plt.show()
```

```python

plt.plot(x,y,linestyle="dashed") 
plt.show()
```

```python
plt.plot(x,y,linestyle="dashed",color="r") 
plt.show()
```

```python
plt.plot(x,y,linestyle="dashed",color="g",linewidth="20") 
plt.show()
```

```python
x1=np.array([8,9,26,40])
y1=np.array([55,45,35,25])

x2=np.array([5,61,26,40])
y2=np.array([60,50,40,30])

plt.plot(x1,y1)
plt.plot(x2,y2)

plt.show()
```

```python
x=np.array([19,40,26,40])
y=np.array([30,40,50,60])
plt.plot(x,y,linestyle="dotted") 
plt.xlabel("age")
plt.ylabel("salary")
plt.title("salary and age relation")
plt.show()
```

```python
x = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90])
y = np.array([120, 140, 135, 160, 180, 200, 195, 220, 250])

plt.plot(x,y)

font2={"family":"serif","color":"green","size":15}

plt.title("Monthly Sales Data",fontdict={"family":"serif","color":"blue","size":20})
plt.xlabel("Months",fontdict=font2)
plt.ylabel("Sales",fontdict=font2)

plt.show()
```
